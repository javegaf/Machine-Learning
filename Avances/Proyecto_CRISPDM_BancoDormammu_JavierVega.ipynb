#### 🧭 **Definición de fases de la metodología Crisp-DM utilizada a lo largo del semestre**

- **Fase 1: Comprensión del negocio (Business Understanding)**

  En esta etapa, se busca comprender los objetivos y requisitos del negocio desde la perspectiva del problema que se está abordando. Se definen los objetivos del proyecto, se establece el marco de trabajo y se desarrolla un plan inicial para alcanzar esos objetivos.

  Aquí, se recopila y se familiariza con los datos disponibles para el proyecto. - **Fase 3: Preparación de los datos (Data Preparation)**

  Durante esta etapa, se limpian y preparan los datos para el modelado. Esto puede incluir la selección de variables relevantes, la transformación de datos, la imputación de valores faltantes, la codificación de variables categóricas y la creación de nuevas características derivadas. El objetivo es garantizar que los datos estén en un formato adecuado y de alta calidad para el análisis posterior.

- **Fase 4: Modelamiento (Modeling)**

  En esta etapa, se seleccionan y se aplican técnicas de modelado para construir y evaluar modelos predictivos o descriptivos. Se pueden probar varias técnicas y algoritmos de modelado, y se ajustan los parámetros de los modelos para optimizar su rendimiento. Esta etapa también implica la evaluación y comparación de los modelos para seleccionar el más adecuado según los criterios definidos anteriormente.

- **Fase 5: Evaluación del modelo (Evaluation)**

  Aquí se evalúan los modelos desarrollados para determinar su idoneidad para abordar los objetivos del negocio. Se utilizan métricas de evaluación adecuadas para medir el rendimiento de los modelos, como precisión, recall, F1-score, entre otras. Además, se puede realizar una validación cruzada o dividir los datos en conjuntos de entrenamiento y prueba para evaluar el rendimiento del modelo en datos no vistos.

- **Fase 6: Despliegue (Deployment)**

  La etapa final implica implementar los resultados del proyecto en el entorno operativo del negocio. Esto puede implicar la integración de los modelos en sistemas existentes, la creación de informes o tableros interactivos, la automatización de procesos o cualquier otra acción necesaria para garantizar que los resultados se utilicen de manera efectiva para generar valor comercial.

### � **Valores nulos y su concentración**

**Distribución de valores nulos y no nulos en el DataFrame**

````python
# 1. Análisis general de nulos
total_celdas = data_frame.size
nulos_totales = data_frame.isnull().sum().sum()
porcentaje_total = (nulos_totales / total_celdas) * 100

# 2. Análisis por columnas
nulos_por_columna = data_frame.isnull().sum()
porcentaje_por_columna = (nulos_por_columna / len(data_frame)) * 100
resumen_nulos = pd.DataFrame({
    'Columna': data_frame.columns,
    'Nulos': nulos_por_columna.values,
    '% Nulos': porcentaje_por_columna.values,
    'Tipo Dato': data_frame.dtypes.values
}).sort_values('% Nulos', ascending=False)

# 3. Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico circular (general)
ax1.pie([nulos_totales, total_celdas - nulos_totales],
        labels=['Nulos', 'No Nulos'],
        colors=['#ff6b6b', '#51cf66'],
        autopct='%1.1f%%',
        explode=(0.1, 0),
        shadow=True)
ax1.set_title('Distribución General de Valores Nulos\n', fontweight='bold')

# Gráfico de barras (top columnas con nulos)
top_nulos = resumen_nulos[resumen_nulos['Nulos'] > 0].head(10)
if not top_nulos.empty:
    bars = ax2.barh(top_nulos['Columna'], top_nulos['% Nulos'],
                   color='#f76707', edgecolor='black')
    ax2.bar_label(bars, fmt='%.1f%%', padding=3, fontsize=10)
    ax2.set_title('Top 10 Columnas con más Valores Nulos\n', fontweight='bold')
    ax2.set_xlabel('Porcentaje de Nulos (%)')
    ax2.grid(axis='x', linestyle='--', alpha=0.6)
else:
    ax2.text(0.5, 0.5, '¡No hay valores nulos!',
            ha='center', va='center', fontsize=14, color='green')
    ax2.axis('off')

plt.tight_layout()
plt.show()

# 4. Resultados en consola
print(f"\nRESUMEN GENERAL:")
print(f"• Celdas totales: {total_celdas:,}")
print(f"• Valores nulos: {nulos_totales:,} ({porcentaje_total:.2f}%)")
print(f"• Columnas con nulos: {len(resumen_nulos[resumen_nulos['Nulos'] > 0])} de {len(data_frame.columns)}")

# 5. Tabla detallada
if not top_nulos.empty:
    print("\nDETALLE POR COLUMNA (ordenado por % de nulos):")
    # Versión simple sin estilo condicional para mayor compatibilidad
    display(resumen_nulos.round({'% Nulos': 2}))

    # Recomendaciones automáticas
    print("\nRECOMENDACIONES:")
    for _, row in top_nulos.iterrows():
        if row['% Nulos'] > 30:
            print(f"- [{row['Tipo Dato']}] '{row['Columna']}': {row['% Nulos']:.1f}% nulos → Considera ELIMINAR o IMPUTAR")
        elif row['% Nulos'] > 10:
            print(f"- [{row['Tipo Dato']}] '{row['Columna']}': {row['% Nulos']:.1f}% nulos → Analiza PATRONES")
        elif row['% Nulos'] > 0:
            print(f"- [{row['Tipo Dato']}] '{row['Columna']}': {row['% Nulos']:.1f}% nulos → Mínimo impacto")
else:
    print("\n¡No se encontraron valores nulos en el dataset!")
````

**Matriz de correlación numérica**

````python
# Filtrar solo las columnas numéricas
df_numericas = data_frame_nuevo.select_dtypes(include=['number'])

# Calcular matriz de correlación
matriz_corr_numerica = df_numericas.corr()

# Visualización en mapa de calor
plt.figure(figsize=(14, 10))
sns.heatmap(matriz_corr_numerica, cmap="coolwarm", annot=False, linewidths=0.5)
plt.title("Matriz de Correlación - Solo Variables Numéricas")
plt.tight_layout()
plt.show()

# Umbrales de correlación
umbral_inferior = 0.7
umbral_superior = 0.95  # Excluir correlaciones casi perfectas

# Crear una máscara para tomar solo la parte superior de la matriz (sin la diagonal)
mask = np.triu(np.ones_like(matriz_corr_numerica, dtype=bool))

# Aplicar la máscara y apilar la matriz
correlaciones_altas = matriz_corr_numerica.where(mask).stack()

# Filtrar las correlaciones entre los umbrales
correlaciones_filtradas = correlaciones_altas[
    (abs(correlaciones_altas) > umbral_inferior) &
    (abs(correlaciones_altas) < umbral_superior)
]

# Mostrar resultados ordenados
correlaciones_ordenadas = correlaciones_filtradas.sort_values(key=abs, ascending=False)

# Visualización de los pares con correlación moderada/alta
print(f"Pares de variables con correlación entre {umbral_inferior} y {umbral_superior}:\n")
print(correlaciones_ordenadas)

# Identificar variables redundantes (correlación > umbral_superior)
variables_redundantes = set()
for col in matriz_corr_numerica.columns:
    correlaciones = matriz_corr_numerica[col][
        (matriz_corr_numerica[col] >= umbral_superior) &
        (matriz_corr_numerica[col] < 1.0)  # Excluir autocorrelación
    ]
    if not correlaciones.empty:
        print(f"\n Variable {col} está casi perfectamente correlacionada con:")
        print(correlaciones)
        variables_redundantes.update(correlaciones.index.tolist())
````

### 🕵️ **Análisis de atípicos**

````python
# Seleccionar columnas numéricas
numeric_cols = data_frame_nuevo.select_dtypes(include=['int64', 'float64']).columns

print("ANÁLISIS DE VALORES ATÍPICOS (OUTLIERS)")
print("RESUMEN DE OUTLIERS DETECTADOS:")

outliers_dict = {}

for col in numeric_cols:
    stats = data_frame_nuevo[col].describe()
    Q1 = stats['25%']
    Q3 = stats['75%']
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data_frame_nuevo[(data_frame_nuevo[col] < lower_bound) | (data_frame_nuevo[col] > upper_bound)]

    if not outliers.empty:
        print(f"\nColumna: {col}")
        print(f"- Rango aceptado: [{lower_bound:.2f}, {upper_bound:.2f}]")
        print(f"- Cantidad de outliers: {len(outliers)}")
        print(f"- Valor mínimo atípico: {outliers[col].min()}")
        print(f"- Valor máximo atípico: {outliers[col].max()}")

        outliers_dict[col] = outliers[[col]]

# Mostrar boxplots solo para columnas que tienen outliers
if outliers_dict:
    print("\nMOSTRANDO BOXPLOTS PARA COLUMNAS CON OUTLIERS...")
    for col in outliers_dict:
        plt.figure(figsize=(10, 4))
        plt.boxplot(data_frame_nuevo[col].dropna(), vert=False)
        plt.title(f'Distribución de {col} (con posibles outliers)')
        plt.xlabel(col)
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.show()
else:
    print("\nNo se detectaron valores atípicos en columnas numéricas.")

# Inicializar listas para variables con y sin outliers
variables_con_outliers = []
variables_sin_outliers = []

# Detectar outliers por columna usando el criterio del IQR
for col in numeric_cols:
    Q1 = data_frame_nuevo[col].quantile(0.25)
    Q3 = data_frame_nuevo[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    hay_outliers = data_frame_nuevo[(data_frame_nuevo[col] < lower_bound) | (data_frame_nuevo[col] > upper_bound)].shape[0] > 0

    if hay_outliers:
        variables_con_outliers.append(col)
    else:
        variables_sin_outliers.append(col)

# Conteo
conteo = pd.Series({
    'Con Outliers': len(variables_con_outliers),
    'Sin Outliers': len(variables_sin_outliers)
})

# Mostrar conteo
print("\nConteo de variables:")
print(conteo)

# Mostrar las variables sin outliers
print("\nVariables sin Outliers:")
print(variables_sin_outliers)

# Gráfico de barras con los conteos
plt.figure(figsize=(6, 4))
ax = conteo.plot(kind='bar', color=['tomato', 'mediumseagreen'])

# Añadir los conteos encima de las barras
for i, v in enumerate(conteo):
    ax.text(i, v + 0.1, str(v), ha='center', va='bottom', fontweight='bold')

plt.title('Variables con y sin Outliers')
plt.ylabel('Cantidad de Variables')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
````

**2. ¿Existe correlación entre la cantidad de transacciones y los pagos efectivos?**

**Objetivo:** Evaluar la relación entre las transacciones realizadas por los clientes y los pagos efectivos para ver si estos están directamente relacionados.

**Variables a analizar:**
- **Transacciones por mes:** Txs_T01, Txs_T02, ..., Txs_T12

- **Pagos por mes:** Pago_T01, Pago_T02, ..., Pago_T12

**Matriz de correlación**

Calcularemos la correlación entre las transacciones y los pagos para cada mes.

````python
# Definir columnas de transacciones y pagos reales
transacciones = ['Txs_T01', 'Txs_T02', 'Txs_T03', 'Txs_T04', 'Txs_T05', 'Txs_T06',
                 'Txs_T07', 'Txs_T08', 'Txs_T09', 'Txs_T10', 'Txs_T11', 'Txs_T12']

pagos_nacionales = ['PagoNac_T01', 'PagoNac_T02', 'PagoNac_T03', 'PagoNac_T04', 'PagoNac_T05', 'PagoNac_T06',
                    'PagoNac_T07', 'PagoNac_T08', 'PagoNac_T09', 'PagoNac_T10', 'PagoNac_T11', 'PagoNac_T12']

pagos_internacionales = ['PagoInt_T01', 'PagoInt_T02', 'PagoInt_T03', 'PagoInt_T04', 'PagoInt_T05', 'PagoInt_T06',
                         'PagoInt_T07', 'PagoInt_T08', 'PagoInt_T09', 'PagoInt_T10', 'PagoInt_T11', 'PagoInt_T12']

# Crear columnas de pagos totales por mes
for i in range(12):
    mes = f"T{str(i+1).zfill(2)}"
    data_frame_nuevo[f'PagoTotal_{mes}'] = pd.to_numeric(data_frame_nuevo[f'PagoNac_{mes}'], errors='coerce') + \
                                            pd.to_numeric(data_frame_nuevo[f'PagoInt_{mes}'], errors='coerce')

# Convertimos las columnas de transacciones a numérico
data_frame_nuevo[transacciones] = data_frame_nuevo[transacciones].apply(pd.to_numeric, errors='coerce')

# Seleccionamos las nuevas columnas de pagos totales
pagos_totales = [f'PagoTotal_T{str(i+1).zfill(2)}' for i in range(12)]

# Calculamos la correlación entre transacciones y pagos
correlacion = data_frame_nuevo[transacciones + pagos_totales].corr()

# Extraemos la submatriz de correlación entre transacciones y pagos
correlacion_tx_pago = correlacion.loc[transacciones, pagos_totales]

# Visualizamos la matriz de correlación
plt.figure(figsize=(10, 8))
sns.heatmap(correlacion_tx_pago, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title('Correlación entre Transacciones y Pagos por Mes')
plt.xlabel('Pagos Totales')
plt.ylabel('Transacciones')
plt.show()
````

**Primera imputación de nulos (mediana, media y/o moda)**

````python
# Copia inicial de data_frame_nuevo
data_frame_limpio = data_frame_nuevo.copy()

# 1. Imputación de Nulos (Mediana/Media)
selected_variables = data_frame_limpio.select_dtypes(include=['float64', 'int64']).columns.tolist()  # Filtrar solo columnas numéricas

# Umbral para alta proporción de nulos
null_threshold = 0.2

# Imputación de nulos en las variables seleccionadas
for feature in selected_variables:
    num_nulls = data_frame_limpio[feature].isnull().sum()  # Contar el número de valores nulos
    kurtosis = data_frame_limpio[feature].kurtosis()  # Curtosis de la variable
    skewness = data_frame_limpio[feature].skew()  # Asimetría de la variable

    total_values = len(data_frame_limpio[feature])  # Total de valores en la variable
    null_proportion = num_nulls / total_values  # Proporción de valores nulos

    # Decidir el método de imputación
    if null_proportion > null_threshold:  # Si la proporción de nulos es mayor que el umbral, usar mediana
        imputation_method = 'mediana'
    elif kurtosis > 0.5 or kurtosis < -0.5 or abs(skewness) > 0.5:  # Si la curtosis o asimetría es alta, usar mediana
        imputation_method = 'mediana'
    else:
        imputation_method = 'media'  # En caso contrario, usar media

    # Realizar la imputación
    if imputation_method == 'mediana':
        data_frame_limpio[feature].fillna(data_frame_limpio[feature].median(), inplace=True)  # Imputar con mediana
    else:
        data_frame_limpio[feature].fillna(data_frame_limpio[feature].mean(), inplace=True)  # Imputar con media

    # Imprimir información sobre la imputación realizada
    print(f"Feature: {feature}")
    print(f"  Nulos: {num_nulls} ({null_proportion:.2%})")
    print(f"  Curtosis: {kurtosis}")
    print(f"  Asimetría: {skewness}")
    print(f"  Método de imputación: {imputation_method}")
    if null_proportion > null_threshold:
        print(f"  Advertencia: {feature} tiene una alta proporción de valores nulos ({null_proportion:.2%})")

# Verificar el DataFrame después de la imputación de nulos
print("Imputación de nulos realizada. Datos limpios:")
print(data_frame_limpio.head())
````

**Observaciones**

---

**¿Por qué imputamos por la mediana y no por la media?**

Tomaremos como ejemplo PagoNac_T12
- Feature: PagoNac_T12
  - Nulos: 8251 (16.14%)
  - Curtosis: 1015.68600096638
  - Asimetría: 21.024207940445738
  - Método de imputación: mediana

Imputamos por la mediana en lugar de la media debido a las siguientes razones:
- Alta curtosis (1015.69): La curtosis extremadamente alta indica una distribución con colas pesadas, lo que sugiere que existen valores extremos que podrían afectar significativamente la media. En distribuciones con alta curtosis, la media no es un buen indicador ya que los valores atípicos distorsionan su valor.
- Alta asimetría (21.02): La asimetría elevada muestra que la distribución de los datos es sesgada. En distribuciones sesgadas, la media se ve afectada por los valores extremos y no refleja correctamente la tendencia central. La mediana, al ser menos sensible a estos valores atípicos, es más robusta y adecuada para la imputación.

**¿Cuándo imputamos por la media?**

Imputamos por la media cuando:
- La distribución de los datos es aproximadamente simétrica (es decir, tiene baja asimetría) y no presenta colas pesadas (curtosis baja).

Ejemplo: Tomemos PagoNac_T13(inventado):
- Nulos: 8251 (16.14%)
- Curtosis: 0.45 (valor bajo)
- Asimetría: 0.03 (valor cercano a 0)

En este caso, imputamos por la media debido a:

- Curtosis baja (0.45): La distribución no tiene colas pesadas, lo que significa que no hay valores atípicos que distorsionen la media.
- Asimetría baja (0.03): La distribución es casi simétrica, lo que indica que la media es una representación fiel de la tendencia central de los datos.

Por lo tanto, la media es adecuada para imputar en distribuciones simétricas con baja curtosis y asimetría.

````python
# Diccionario base para variables mensuales
prefijos = {
    'Fac_T': 'Facturación_Total',
    'FacCI_T': 'Facturación_Crédito_Inmediato',
    'FacCN_T': 'Facturación_Crédito_No_Revolvente',
    'FacAI_T': 'Facturación_Avances_Internacionales',
    'FacAN_T': 'Facturación_Avances_Nacionales',
    'FacPAT_T': 'Facturación_PAT',
    'Txs_T': 'Transacciones_Totales',
    'TxsCI_T': 'Transacciones_Crédito_Inmediato',
    'TxsAI_T': 'Transacciones_Avances_Internacionales',
    'PagoInt_T': 'Pagos_Internacionales',
    'PagoNac_T': 'Pagos_Nacionales'
}

# Meses en español
meses = {
    '01': 'Enero', '02': 'Febrero', '03': 'Marzo', '04': 'Abril',
    '05': 'Mayo', '06': 'Junio', '07': 'Julio', '08': 'Agosto',
    '09': 'Septiembre', '10': 'Octubre', '11': 'Noviembre', '12': 'Diciembre'
}

# Crear diccionario de reemplazo para columnas
nuevos_nombres = {}

for col in data_frame_limpio.columns:
    for prefijo, descripcion in prefijos.items():
        if col.startswith(prefijo):
            mes = col.replace(prefijo, '')  # Extraer el número del mes
            if mes in meses:
                nuevos_nombres[col] = f"{descripcion}_{meses[mes]}"

# Aplicar renombramiento solo a variables mensuales
data_frame_renombrado = data_frame_limpio.rename(columns=nuevos_nombres)

# Mostrar ejemplo
print("Ejemplo de columnas renombradas:")
for original, nuevo in list(nuevos_nombres.items())[:10]:
    print(f"{original} → {nuevo}")
````

**Variables Cuantitativas v/s Cualitativas**

````python
# Definir listas para almacenar nombres de variables en cada categoría
variables_cuantitativas_discretas = []
variables_cuantitativas_continuas = []
variables_cualitativas_ordinal = []
variables_cualitativas_nominal = []

# Iterar sobre las columnas del DataFrame
for columna in data_frame_renombrado.columns:
    tipo_dato = data_frame_renombrado[columna].dtype

    # Cuantitativa discreta
    if tipo_dato == 'int64' or tipo_dato == 'float64':
        if data_frame_renombrado[columna].nunique() <= 15:  # Se elige un umbral arbitrario de 15 para distinguir variables discretas de continuas
            variables_cuantitativas_discretas.append(columna)
        else:
            variables_cuantitativas_continuas.append(columna)

    # Cualitativa ordinal
    elif tipo_dato == 'object':
        # Se asume que si una columna tiene menos de 15 valores únicos, es ordinal, de lo contrario, es nominal
        if data_frame_renombrado[columna].nunique() <= 15:
            variables_cualitativas_ordinal.append(columna)
        else:
            variables_cualitativas_nominal.append(columna)

# Imprimir variables cuantitativas discretas
print("Variables cuantitativas discretas presentes en el conjunto de datos:")
for feature in variables_cuantitativas_discretas:
    print("\t", feature)
print("\nTotal de variables cuantitativas discretas:", len(variables_cuantitativas_discretas))

# Imprimir variables cuantitativas continuas
print("\nVariables cuantitativas continuas presentes en el conjunto de datos:")
for feature in variables_cuantitativas_continuas:
    print("\t", feature)
print("\nTotal de variables cuantitativas continuas:", len(variables_cuantitativas_continuas))

# Imprimir variables cualitativas ordinales
print("\nVariables cualitativas ordinales presentes en el conjunto de datos:")
for feature in variables_cualitativas_ordinal:
    print("\t", feature)
print("\nTotal de variables cualitativas ordinales:", len(variables_cualitativas_ordinal))

# Imprimir variables cualitativas nominales
print("\nVariables cualitativas nominales presentes en el conjunto de datos:")
for feature in variables_cualitativas_nominal:
    print("\t", feature)
print("\nTotal de variables cualitativas nominales:", len(variables_cualitativas_nominal))
````

**¿Por qué se elige un umbral como 15?**

Cuando clasificamos variables cuantitativas en discretas o continuas, la idea es distinguir entre:

- Discretas: toman un conjunto limitado de valores posibles (ej: número de hijos, cantidad de productos comprados).

- Continuas: pueden tomar infinitos valores en un rango, como ingresos, temperatura o tiempo.

Como en los datasets reales muchas veces no sabemos de antemano el tipo semántico de una variable, se usa una heurística:

```
Si una variable numérica tiene ≤ 15 valores únicos distintos, probablemente sea discreta.
Si tiene más de 15 valores únicos, seguramente sea continua.
```



**¿Por qué 15 y no 10 o 20?**

No hay una regla estricta. Se basa en la intuición estadística:

- Menos de 15: usualmente es algo como "niveles", "categorías codificadas como números", conteos simples.

- Más de 15: difícil que sea un conteo simple, y más probable que sea una medición más granular o continua.
